```python
import gridsort as grid
import griddemo as demo

import time
```

# Config: 
sample a (N, D) point cloud on a given backend

```python
D = 2
N = 128**D
gridshape = (128,)*D

backend = "numpy" #or "torch" or "cupy"

X = demo.sample(method="ball", size=(N, D), backend=backend)
print("initial point cloud X has shape", X.shape)
print("points are randomly stored in memory. First five points:\n")
for point in X[:5]:
    x, y = point
    print(f"({x:.3f}, {y:.3f})")
```

# Sort the cloud

```python
start = time.time()

#level is an important parameter, 
#controlling the speed-quality tradeoff of the sort. 
# Try level = 1, 2, 3: 
# level = 1 sort only cartesian axes, 
# level = 2 add diagonals, 
# level = 3 add tridiagonals (for >=3D point clouds only).
level = 2 

order = grid.argsort(X, gridshape, level = level)
Y = X[order]

print(f"elapsed: {(time.time() - start):.2f} s")
```

# Build the grid

```python
print(f"Y has same shape {Y.shape} as X, we just permuted the points,")
Ygrid = Y.reshape(*gridshape, D)
print(f"but now the storage is consistant, following a grid layout with shape {Ygrid.shape}.")

demo.print_zoom(Ygrid, center = [42, 42])
demo.print_zoom(Ygrid, center = [73, 21])
```

# Show the result
- plot the point cloud and the mesh formed by the rows + columns of the grid
- print the neighbormap. Neighbormap counts the number of nearest neighbor  \
found at a given relative grid offset

```python
demo.show_result(Ygrid)
```

# 3D version

```python
D = 3
N = 32**3
gridshape = (32,)*D

X = demo.sample(method="square", size=(N, D), backend=backend)
print(f"X has shape {X.shape}")
Y = grid.sort(X, gridshape = gridshape, level = 3)
Ygrid = Y.reshape(*gridshape, D)
```

```python
demo.show_result(Ygrid)
```
